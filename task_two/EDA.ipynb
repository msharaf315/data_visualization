```python
def _get_alphabet_inbetween_two_letters(start_letter, end_letter):
    return list(map(chr, range(ord(start_letter)+1, ord(end_letter))))
    
```

```python
def _get_letter_and_number(code):
    start_letter = code[0]
    start_number = code[1:]
    return start_letter, int(start_number)
```

```python
def _handle_code_with_dash(detail_code, code_disease_dict, cause):
    dash_separated = detail_code.split("-")
    start_code = dash_separated[0].strip()
    end_code = dash_separated[1].strip()
    start_letter, start_number = _get_letter_and_number(start_code)
    end_letter, end_number = _get_letter_and_number(end_code)
    if not (start_letter == end_letter):
        letters_list = _get_alphabet_inbetween_two_letters(start_letter, end_letter)
        # Handle start letter letter plus first number till 99
        for i in range(int(start_number), 100):
            code_number = i 
            if code_number <10:
                code_number = f"0{code_number}"
            new_code = f"{start_letter}{code_number}"
            code_disease_dict[new_code] = cause
        # Handle in between letters
        for letter in letters_list:
            for i in range(0,100):
                code_number = i 
                if code_number <10:
                    code_number = f"0{code_number}"
                new_code = f"{letter}{code_number}"
                code_disease_dict[new_code] = cause
        # Handle end letter 
        for i in range(0, int(end_number)+1):
            code_number = i 
            if code_number <10:
                code_number = f"0{code_number}"
            new_code = f"{end_letter}{code_number}"
            code_disease_dict[new_code] = cause
             
    if start_letter == end_letter:
        for i in range(int(start_number), int(end_number) + 1):
            code_number = i 
            if code_number <10:
                code_number = f"0{code_number}"
            new_code = f"{start_letter}{code_number}"
            code_disease_dict[new_code] = cause
            # print(f'adding {new_code}:{cause}')
```

```python
def _handle_single_code(detail_code, code_disease_dict, cause):
    detail_code = detail_code.strip()
    if len(detail_code) == 3:
        code_disease_dict[detail_code] = cause
        # print(f'adding {detail_code}:{cause}')
    else:
        print(f"error single code has more than 3 letters {detail_code}")
```

```python
code_disease_dict = {}

for _, row in df_cause_of_death_code.iterrows():
    detailed_code_full = row["Detailed List Numbers"]
    cause = row["Cause"]
    # print(detailed_code_full)
    if not pd.isna(detailed_code_full):
        # Hanlde comma separated codes
        if "," in detailed_code_full:
            comma_separated = detailed_code_full.split(",")
            for detail_code in comma_separated:
                if "-" in detail_code:
                    _handle_code_with_dash(detail_code, code_disease_dict, cause)
                else:
                    _handle_single_code(detail_code, code_disease_dict, cause)
        # handle - separated codes
        elif  "-" in detailed_code_full:
                _handle_code_with_dash(detailed_code_full, code_disease_dict, cause)
        # handle single codes
        else: 
            _handle_single_code(detailed_code_full, code_disease_dict, cause)
for key, value in code_disease_dict.items():
    print(key,value)
    df_cause_of_death_code = pd.concat([df_cause_of_death_code, pd.DataFrame([[key, None, value]], columns=df_cause_of_death_code.columns)], ignore_index=True)


```
